```python
valid_gratings = gratings_df[((gratings_df['orientation'].isna()) & (gratings_df['temporal_frequency'].isna())) == False]
print(len(valid_gratings))
```

```python
print((valid_gratings['orientation']).value_counts())
```

```python
valid_gratings['Duration'] = valid_gratings['stop_time'] - valid_gratings['start_time']
print(valid_gratings['Duration'])
```

```python
print("Duration column: ")
print(np.mean(valid_gratings['Duration']))
print(np.min(valid_gratings['Duration']))
print(np.max(valid_gratings['Duration']))
print(np.std(valid_gratings['Duration']))
```

```python
valid_gratings['Gap'] = valid_gratings['start_time'].shift(-1) - valid_gratings['stop_time']
print(valid_gratings['Gap'].head())
print(len(valid_gratings[valid_gratings['Gap']>0]))
print(len(valid_gratings[valid_gratings['Gap']==0]))
print(len(valid_gratings[valid_gratings['Gap']<0]))
```

```python
gap_min = np.min(valid_gratings['Gap'])
gap_max = np.max(valid_gratings['Gap'])
gap_mean = np.mean(valid_gratings['Gap'])
gap_median = valid_gratings['Gap'].median()
gap_std = np.std(valid_gratings['Gap'])
print(gap_min,gap_max, gap_mean, gap_median, gap_std)
```
